```python
import pandas as pd
import numpy as np
```

# **1. Coleta de dados**

```python
# Lendo o arquivo CSV com a codificação correta
df = pd.read_csv('vendas.csv', sep=';')

print("Dados originais:")
print(df)
print("\nInformações do DataFrame:")
print(df.info())
```

# **2. Processamento de dados**

```python
# Convertendo a coluna 'data' para o tipo datetime
df['data'] = pd.to_datetime(df['data'])

# Preenchendo valores ausentes na coluna 'quantidade' com a mediana
df['quantidade'] = df['quantidade'].fillna(df['quantidade'].median())

# Criando uma nova coluna 'valor_total' multiplicando 'quantidade' por 'preco'
df['valor_total'] = df['quantidade'] * df['preco']

print("\nDados após processamento:")
print(df)
```

# **3. Limpeza de dados**

```python
# Removendo linhas duplicadas (se houver)
df.drop_duplicates(inplace=True)

# Removendo outliers da coluna 'preco' (considerando outliers valores acima de 1000)
df = df[df['preco'] <= 1000]

# Arredondando valores numéricos
df = df.round({'preco': 2, 'valor_total': 2})

print("\nDados após limpeza:")
print(df)
print("\nInformações do DataFrame final:")
print(df.info())

# Salvando os dados processados e limpos em um novo arquivo CSV
df.to_csv('vendas_processadas.csv', index=False)
print("\nDados processados e limpos salvos em 'vendas_processadas.csv'")
```
