```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
```

```python
# Definir o número de registros
n_registros = 10000

# Criar dados aleatórios
np.random.seed(42)  # Para reprodutibilidade
# Datas de venda
data_inicio = datetime(2022, 1, 1)
datas_venda = [data_inicio + timedelta(days=i) for i in range(n_registros)]

# IDs de cliente
ids_cliente = np.random.randint(1, 1001, n_registros)

# Produtos e categorias
produtos = ['Laptop', 'Smartphone', 'Tablet', 'Fones de ouvido', 'Smartwatch', 'Câmera', 'TV', 'Console de jogos', 'Impressora', 'Roteador']
categorias = ['Eletrônicos', 'Eletrônicos', 'Eletrônicos', 'Acessórios', 'Wearables', 'Eletrônicos', 'Eletrônicos', 'Jogos', 'Escritório', 'Redes']

indices_produto = np.random.randint(0, len(produtos), n_registros)
produtos_vendidos = [produtos[i] for i in indices_produto]
categorias_vendidas = [categorias[i] for i in indices_produto]

# Quantidades e valores
quantidades = np.random.randint(1, 6, n_registros)
valores_unitarios = np.random.uniform(50, 2000, n_registros)
valores_venda = quantidades * valores_unitarios

# Estados
estados = ['CA', 'NY', 'TX', 'FL', 'IL', 'PA', 'OH', 'GA', 'NC', 'MI']
estados_venda = np.random.choice(estados, n_registros)

# Idade dos clientes
idades_cliente = np.random.randint(18, 80, n_registros)
```

```python
# Criar DataFrame
df = pd.DataFrame({
    'id_pedido': range(1, n_registros + 1),
    'data_venda': datas_venda,
    'id_cliente': ids_cliente,
    'idade_cliente': idades_cliente,
    'produto': produtos_vendidos,
    'categoria': categorias_vendidas,
    'quantidade': quantidades,
    'valor_unitario': valores_unitarios,
    'valor_venda': valores_venda,
    'estado': estados_venda
})

# Adicionar feedbacks mais variados e realistas
feedbacks_positivos = [
    "Excelente produto! Superou minhas expectativas.",
    "Entrega rápida e produto de ótima qualidade.",
    "Muito satisfeito com a compra, recomendo!",
    "Atendimento ao cliente excepcional.",
    "Produto com ótimo custo-benefício.",
    "Fiquei impressionado com a qualidade do produto.",
    "A entrega foi mais rápida do que o esperado.",
    "Produto exatamente como descrito, muito bom!",
    "Ótima experiência de compra, voltarei a comprar.",
    "O produto chegou em perfeitas condições."
]

feedbacks_neutros = [
    "Produto ok, atendeu às expectativas.",
    "Entrega dentro do prazo previsto.",
    "Qualidade do produto é razoável.",
    "Nada extraordinário, mas não tenho reclamações.",
    "Produto cumpre o que promete.",
    "Atendimento padrão, sem problemas.",
    "Preço justo pelo que oferece.",
    "Produto funciona bem, mas poderia ser melhor.",
    "Entrega normal, sem atrasos ou adiantamentos.",
    "Compra tranquila, sem surpresas."
]

feedbacks_negativos = [
    "Produto chegou com defeito, muito decepcionado.",
    "Entrega atrasada, péssimo serviço.",
    "Qualidade inferior ao anunciado.",
    "Péssimo atendimento ao cliente.",
    "Produto caro para o que oferece.",
    "Arrependido da compra, não recomendo.",
    "O produto não funcionou como esperado.",
    "Embalagem danificada, produto chegou quebrado.",
    "Demorou muito para entregar e ainda veio errado.",
    "Experiência de compra frustrante, não comprarei novamente."
]
```

```python
# Gerar feedbacks aleatórios com a distribuição desejada
feedbacks = np.random.choice([0, 1, 2], n_registros, p=[0.6, 0.3, 0.1])
df['feedback'] = [
    np.random.choice(feedbacks_positivos) if f == 0 else
    np.random.choice(feedbacks_neutros) if f == 1 else
    np.random.choice(feedbacks_negativos)
    for f in feedbacks
]
```

```python
# Salvar o DataFrame como CSV
df.to_csv('vendas_ecommerce.csv', index=False)

print("Arquivo 'vendas_ecommerce.csv' gerado com sucesso!")
```
